câu 4:Thống kê dữ liệu thiếu cho cột dân tộc và hiệu chỉnh dữ liệu thiếu như sau: Mặc định thiếu 
thì điền giá trị 0.

```python
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

print("Số lượng dữ liệu thiếu ở cột DT:")
print(df["DT"].isnull().sum())

df["DT"] = df["DT"].fillna(0)

print("\nSau khi hiệu chỉnh:")
print(df["DT"].isnull().sum())
```

Câu 5:Thống kê dữ liệu thiếu cho biến T1 và hiệu chỉnh dữ liệu, lưu ý việc thay thế dữ liệu thiếu sử 
dụng phương pháp Mean.

```python
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

print("5 dòng đầu dữ liệu")
print(df.head())

missing_count = df["T1"].isnull().sum()

print("\nSố lượng dữ liệu thiếu của T1:", missing_count)

mean_value = df["T1"].mean()

print("Giá trị trung bình của T1:", mean_value)

df["T1"].fillna(mean_value, inplace=True)

missing_after = df["T1"].isnull().sum()

print("\nSố lượng dữ liệu thiếu sau khi hiệu chỉnh:", missing_after)

df.to_csv("dulieuxettuyendaihoc_fixed.csv", index=False)

print("\nĐã lưu file: dulieuxettuyendaihoc_fixed.csv")
```

Câu 6: Xử lý lần lượt tất cả dữ liệu thiếu cho các biến điểm số còn lại

```python
import pandas as pd

df = pd.read_csv("dulieuxettuyendaihoc.csv")

print("Số lượng dữ liệu thiếu trước khi xử lý:\n")
print(df.isnull().sum())

# Danh sách các cột điểm số cần xử lý
cot_diem = ["T1", "L1", "H1", "S1", "V1", "X1", "D1", "N1"]

# Thay thế dữ liệu thiếu bằng Mean
for cot in cot_diem:
    if cot in df.columns:
        mean_value = df[cot].mean()
        df[cot].fillna(mean_value, inplace=True)

print("\nSố lượng dữ liệu thiếu sau khi xử lý:\n")
print(df[cot_diem].isnull().sum())

# Lưu file mới
df.to_csv("dulieuxettuyendaihoc_daxuly.csv", index=False)

print("\nĐã xử lý xong và lưu file mới thành công!")
```

10, 11, 12 như sau:

```python
def minmax_to_4(series):
    return 4 * (series - series.min()) / (series.max() - series.min())

df['US_TBM1'] = minmax_to_4(df['TBM1'])
df['US_TBM2'] = minmax_to_4(df['TBM2'])
df['US_TBM3'] = minmax_to_4(df['TBM3'])
```
