```
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Tải thêm gói dữ liệu còn thiếu để chạy hàm word_tokenize
nltk.download('punkt_tab')
nltk.download('stopwords')

text = "This is a sample sentence showing how to remove stopwords using NLTK."
stop_words = set(stopwords.words('english'))

# Tách từ (Bây giờ sẽ chạy mượt mà không lỗi nữa)
words = word_tokenize(text)

# Câu 4 & 5: Lọc bỏ các từ nằm trong danh sách stopword
filtered_text = [w for w in words if w.lower() not in stop_words]

print("Văn bản gốc:", words)
print("Văn bản sau khi loại bỏ stopword:", filtered_text)
```

```
from nltk.corpus import wordnet

word = "computer"
synsets = wordnet.synsets(word)

if synsets:
    first_syn = synsets[0]
    print(f"Từ khóa: {word}")
    print("Định nghĩa:", first_syn.definition())
    print("Ví dụ thực tế:", first_syn.examples())
else:
    print("Không tìm thấy từ này trong WordNet.")
```

```
from nltk.corpus import wordnet

word = "good"
synonyms = []
antonyms = []

for syn in wordnet.synsets(word):
    for l in syn.lemmas():
        synonyms.append(l.name())
        if l.antonyms():
            antonyms.append(l.antonyms()[0].name())

print(f"Từ đồng nghĩa của '{word}':", set(synonyms))
print(f"Từ trái nghĩa của '{word}':", set(antonyms))
```

```
from nltk.corpus import wordnet

# Lấy nghĩa danh từ (pos='n') của 2 từ
noun1 = wordnet.synset('dog.n.01')
noun2 = wordnet.synset('cat.n.01')

# Tính độ tương đồng Wu-Palmer (thang điểm từ 0 đến 1)
similarity = noun1.wup_similarity(noun2)
print(f"Độ tương đồng giữa 'dog' và 'cat' (Danh từ): {similarity:.4f}")
```

```
from nltk.corpus import wordnet

# Lấy nghĩa động từ (pos='v') của 2 từ
verb1 = wordnet.synset('run.v.01')
verb2 = wordnet.synset('walk.v.01')

similarity = verb1.wup_similarity(verb2)
print(f"Độ tương đồng giữa 'run' và 'walk' (Động từ): {similarity:.4f}")
```

```
import random
from nltk.corpus import names

# --- CÂU 12 ---
# Tạo danh sách các tuple (tên, nhãn)
labeled_names = ([(name, 'male') for name in names.words('male.txt')] +
                 [(name, 'female') for name in names.words('female.txt')])

# Trộn ngẫu nhiên danh sách
random.seed(42) # Giữ kết quả cố định khi chạy lại
random.shuffle(labeled_names)

print("15 kết hợp ngẫu nhiên đầu tiên:")
for name, label in labeled_names[:15]:
    print(f"- {name}: {label}")


# --- CÂU 13 ---
# Định nghĩa hàm trích xuất chữ cái cuối cùng làm đặc trưng (Feature)
def gender_features(name):
    return {'last_letter': name[-1].lower()}

# Trích xuất ký tự cuối cùng và tạo mảng mới gồm (đặc trưng, nhãn)
featuresets = [(gender_features(name), gender) for (name, gender) in labeled_names]

print("\n5 phần tử đầu tiên trong mảng đặc trưng mới:")
for item in featuresets[:5]:
    print(item)
```
